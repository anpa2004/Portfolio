#### Problem 2
##### Numerically Graphing Solutions to 1D Heat Equation

```python
import numpy as np
import plotly.graph_objects as go

k = 2.37 # W/cmK
rho = 2.7 #g/cm^3
c = 0.897 #J/gK
l = 10 #cm
m = 100 # num points

# Lists of data
x_list = np.linspace(0,l,m)
t_list = np.linspace(0,20,m)
u_list = np.empty((m,m))

# Function to solve heat equation
def u(x,t,k,rho,c,theta):
    """  
    Solution to the 1D heat equation given values of x,t and theta, 
    satisfying a dirichlet boundary condition and physical properties
    rho, c, k
    """
    return np.exp(-k*theta**2*t/(rho*c))*np.sin(theta*x)

# Iterating through values to solve
for i in range(len(x_list)):
    for j in range(len(t_list)):
        x = x_list[i]
        t = t_list[j]
        u_list[i,j] = u(x,t,k,rho,c,theta=np.pi/l)

# Creating contour plot
fig = go.Figure(data=[go.Contour(
    z=np.linalg.matrix_transpose(u_list),
    x=x_list, # sets the x coordinates
    y=t_list, # sets the y coordinates # customize the color scale
    contours=dict(
        showlabels=False, # show labels on contour lines
        coloring='heatmap', # use heatmap gradient coloring between lines
    ),
    colorbar=dict(
        title='Temperature (K)'
    )
)])

fig.update_layout(title=r'Heat Equation Solution', title_x=0.5)
fig.update_xaxes(title_text='x (cm)')
fig.update_yaxes(title_text='t (sec)')
fig.show()

# Creating 3D plot
fig = go.Figure(
    data=[go.Surface(x=x_list, y=t_list, z=np.transpose(u_list))]
)

fig.update_layout(
    title=dict(text='Heat Equation Solution'),
    autosize=False,
    width=700,
    height=700,
    margin=dict(l=65, r=50, b=65, t=90),
    scene=dict(
        xaxis_title='x (cm)',
        yaxis_title='t (sec)',
        zaxis_title='u(x,t)'
    )
)

fig.show()
```

#### Problem 3
##### Centered Difference Approximations of 2nd Derivatives

```python
import numpy as np

# Centered difference approximations
def c1h(f,x0,h):
    """
    A centered difference approximation of f''(x_0)

    Args: 
        f: function to approximate the difference of
        x0: Point at which to evaluate
        h: limit parameter in limit definition of the derivative
    """
    return (f(x0+h) - 2*f(x0) + f(x0-h))/(h**2)

def c2h(f,x0,h):
    """
    A centered difference approximation of f''(x_0)

    Args: 
        f: function to approximate the difference of
        x0: Point at which to evaluate
        h: limit parameter in limit definition of the derivative
    """
    return (-f(x0+2*h) + 16*f(x0+h) - 30*f(x0) + 16*f(x0-h) - f(x0 - 2*h))/(12*h**2)

# Calculating the derivative of f(x)=e^x at x=7pi/8
f = lambda x: np.exp(x)
x0 = 7*np.pi/8
h=0.1*2**(-np.linspace(0,16,17))
c1h_sol = c1h(f,x0,h)
c2h_sol = c2h(f,x0,h)

# True Solution, error calculation
truth = np.exp(7*np.pi/8)
log_h = [np.log10(hval) for hval in h]

diff1 = [truth - c for c in c1h_sol]
log_diff1 = [np.log10(np.abs(t)) for t in diff1]
diff2 = [truth - c for c in c2h_sol]
log_diff2 = [np.log10(np.abs(t)) for t in diff2]


# Creating Plots
trace1 = go.Scatter(x=np.flip(log_h), y=log_diff1, mode='lines+markers', name='C1(h)')
trace2 = go.Scatter(x=np.flip(log_h), y=log_diff2, mode='lines+markers', name='C2(h)')

# Create the Figure object and add the trace
fig = go.Figure(data=[trace1,trace2])

# Customize the layout (optional)
fig.update_layout(title='Error Between Solution and Truth',
                  xaxis_title='log10(h)',
                  yaxis_title='log10(Error)')

# Display the figure (e.g., in a Jupyter environment or browser)
fig.show()

# Finding the slopes of the lines before issues
m1 = (log_diff1[1] - log_diff1[0])/(log_h[1]-log_h[0])
m2 = (log_diff2[1] - log_diff2[0])/(log_h[1]-log_h[0])
print(f'log-log slope for C1(h) = {m1}')
print(f'log-log slope for C2(h) = {m2}')
```
