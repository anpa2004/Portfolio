```python
import numpy as np
from functions import ivp_solver
import plotly.graph_objects as go 

def f(t,y):
    return y**2/(1+t)

def f_truth(t):
    return - 1/np.log(t+1)

y0 = -1/np.log(2)

t0 = 1
tf = 2
t_eval = [1,1.2,1.4,1.6,1.8,2]


# Solve using Forward Euler, Dt = 0.025 ################
dt = 0.025
t_fe,y_fe,n_fe = ivp_solver(f,t0,y0,tf,dt,method='forward_euler',solver = 'newton',fprime=None,tol=1e-8,nmax=100,neval=True)
# error
t_truth = np.linspace(t0,tf,len(t_fe))
y_truth = f_truth(t_truth)
error_fe = y_fe - y_truth
eu_sols = []
for t in t_eval:
    i = np.where(t_fe == t)
    eu_sols.append(error_fe[i])
true_solution = go.Scatter(x=t_truth,y=y_truth,mode='lines',name='True Solution')

# Solve using Explicit Trapezoidal, Dt = 0.5 ################
dt = 0.05
t_tr,y_tr,n_tr = ivp_solver(f,t0,y0,tf,dt,method='trapezoidal',solver = 'newton',fprime=None,tol=1e-8,nmax=100,neval=True)
# error
t_truth = np.linspace(t0,tf,len(t_tr))
y_truth = f_truth(t_truth)
error_tr = y_tr - y_truth
tr_sols = []
for t in t_eval:
    i = np.where(t_tr == t)
    tr_sols.append(error_tr[i])

# Solve using Explicit Trapezoidal, Dt = 0.1 ################
dt = 0.1
t_rk,y_rk,n_rk = ivp_solver(f,t0,y0,tf,dt,method='rk4',neval=True)
t_rk = np.array([round(t,2) for t in t_rk])
# error
t_truth = np.linspace(t0,tf,len(t_rk))
y_truth = f_truth(t_truth)
error_rk = y_rk - y_truth
rk_sols = []
for t in t_eval:
    i = np.where(t_rk == t)
    rk_sols.append(error_rk[i])

fig = go.Figure()
numerical_solution = go.Scatter(x=t_fe,y=y_fe,mode='lines',name='Forward Euler')
numerical_solution2 = go.Scatter(x=t_tr,y=y_tr,mode='lines',name='Trapezoidal')
numerical_solution3 = go.Scatter(x=t_rk,y=y_rk,mode='lines',name='RK4')

fig.add_trace(numerical_solution)
fig.add_trace(numerical_solution2)
fig.add_trace(numerical_solution3)
fig.add_trace(true_solution)
fig.update_layout(title='Solution to IVP- Real and Simulated',xaxis_title='t',yaxis_title='y')
fig.show()

fig = go.Figure()
fe_error = go.Scatter(x=t_fe,y=np.log10(np.abs(error_fe)),mode='lines',name='Forward Euler Error')
tr_error = go.Scatter(x=t_tr,y=np.log10(np.abs(error_tr)),mode='lines',name='Trapezoidal Error')
fig.add_trace(fe_error)
rk_error = go.Scatter(x=t_rk,y=np.log10(np.abs(error_rk)),mode='lines',name='RK4 Error')
fig.add_trace(rk_error)
fig.add_trace(tr_error)
fig.update_layout(title='Solution Error',xaxis_title='t',yaxis_title='y')
fig.show()

# Creating a latex table
for i in range(len(t_eval)):
    print(f'{t_eval[i]} & {eu_sols[i][0]:.3E} & {tr_sols[i][0]:.3E} & {rk_sols[i][0]:.3E}\\\\')
print(f'{n_fe} & {n_tr} & {n_rk}')
print(f'{len(t_fe)} & {len(t_tr)} & {len(t_rk)}')
```
