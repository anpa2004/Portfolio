#### Problem 1
##### Forward Euler integration

```python
import numpy as np
import plotly.graph_objects as go 
from functions import forward_euler

# Forward Euler Function
def f(t,y):
    yp = (1+t)/(1+y)
    return yp

def f_truth(t):
    """  
    True solution to compare with integrated solution
    """
    return np.sqrt(t**2 + 2*t + 6) - 1

# Integration Paramters
t0 = 1
dt = 0.5
tf = 2
y0 = 2

# Solving Equation
t_vec,y_vec = forward_euler(f,t0,y0,tf,dt)
truth = f_truth(t_vec)

# Plotting
fig = go.Figure()
numerical_solution = go.Scatter(x=t_vec,y=y_vec,mode='lines',name='Numerical Solution')
true_solution = go.Scatter(x=t_vec,y=truth,mode='lines',name='True Solution')
fig.add_trace(numerical_solution)
fig.add_trace(true_solution)
fig.update_layout(title='Solution to IVP- Real and Simulated',xaxis_title='t',yaxis_title='y')
fig.show()

# Error Analysis
error = y_vec - truth
log_error = [np.log10(np.abs(err)) for err in error]

fig = go.Figure()
error_trace = go.Scatter(x=t_vec,y=log_error,mode='lines',name='Log(Error)')
fig.add_trace(error_trace)
fig.update_layout(title='Log(Error) for Numerical Approximation',xaxis_title='t',yaxis_title='log(|Y(t)-y_k(t)|)')
fig.show()

print(f'Error at t=tf = {error[-1]}')
```

#### Problem 2
##### Error due to $\Delta t$ size

```python
import numpy as np
from functions import *
import plotly.graph_objects as go 

# Forward Euler Function
def f(t,y):
    yp = (1+t)/(1+y)
    return yp

def f_truth(t):
    """  
    True solution to compare with integrated solution
    """
    return np.sqrt(t**2 + 2*t + 6) - 1

# Integration Paramters
t0 = 1
dt_list = 10**(-np.linspace(5,10,6))
tf = t0+1e-5
y0 = 2
t_truth = np.linspace(1,tf,100)

# Solving Equation
t_vec_list = []
y_vec_list = []
for dt in dt_list:
    t_vec,y_vec = forward_euler(f,t0,y0,tf,dt)
    t_vec_list.append(t_vec)
    y_vec_list.append(y_vec)
    print(t_vec[-1] - 1+1e-5)
truth = f_truth(t_truth)

# Plotting
fig = go.Figure()
names = ['1e-5','1e-6','1e-7','1e-8','1e-9','1e-10']
for i in range(len(y_vec_list)):
    y = y_vec_list[i]
    t = t_vec_list[i]
    nm = names[i]
    numerical_solution = go.Scatter(x=t,y=y,mode='lines',name=nm)
    fig.add_trace(numerical_solution)
true_solution = go.Scatter(x=t_truth,y=truth,mode='lines',name='True Solution')
fig.add_trace(true_solution)
fig.update_layout(title='Solution to IVP- Real and Simulated for Different dt',xaxis_title='t',yaxis_title='y')
fig.show()

# # Error Analysis
y_end = []
for i in range(len(y_vec_list)):
    y_end.append(y_vec_list[i][-1])

y_truth = f_truth(tf)

error = [np.log10(np.abs(y - y_truth)) for y in y_end]

fig = go.Figure()
trace = go.Scatter(x=np.log10(dt_list),y=error,mode='lines',name='Error')
fig.add_trace(trace)
fig.update_layout(title='Error vs dt',xaxis_title='log10(dt)',yaxis_title='log10(error)')
fig.update_xaxes(autorange='reversed')
fig.show()
```
