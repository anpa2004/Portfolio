```python
import numpy as np

# Problem 1
# Givens
F = 100000 #lbf
h = 80 #kft
Tc = 4840 #R
At = 0.5 #ft^2
Isp = 400 #s
gamma = 1.26 
Gamma = 0.6599
R = 173.6 #lbf ft/lbm R
g0 = 32.174
gc = g0

mdot = (Isp/F)**-1
print(f'mdot = {round(mdot,4)}')

cstar = np.sqrt(R*gc*Tc)/Gamma
print(f'c* = {round(cstar,4)}')

Pc = mdot*np.sqrt(Tc*R/gc)/(Gamma*At)
print(f'Pc = {round(Pc)} lbf/ft^2')

CF = F/(Pc*At)
print(f'CF = {round(CF,4)}')

C = CF*cstar
print(f'C = {round(C,4)}')

PePc = 2116.2*0.02765/Pc
epsilon = Gamma/np.sqrt(2*gamma/(gamma-1)*(PePc**(2/gamma) - PePc**((gamma+1)/gamma)))
print(f'Pe/Pc = {round(PePc,4)}, Ae/At = {round(epsilon,4)}')

Ae = epsilon*At
De = np.sqrt(4*Ae/np.pi)
print(f'De = {round(De,4)}')

PaPc = 2116.2/Pc
CFi = Gamma*np.sqrt(2*gamma/(gamma-1)*(1-(PePc)**((gamma-1)/gamma)))+(PePc - PaPc)*epsilon
print(f'Cfi = {round(CFi,4)}')
Fsl = CFi*Pc*At
print(f'sea level Fi = {round(Fsl,4)} lbf')
```

```python
from scipy.optimize import root_scalar
import numpy as np

# givens
OF = 2
F = 3000000 # lbf
tb = 120 # s
Dg = 150 # in
Pc = 700 # psi
Pa = 14.7 # psi
rdot = 0.03 # in/s
cstar = 5912 # ft/s
etac = 0.95
gamma = 1.152
Gamma = 0.6386
epsilon = 7.72
gc = 32.174

func = lambda M: -M + 1/epsilon*(2/(gamma+1)*(1+(gamma-1)/2*M**2))**((gamma+1)/2*(gamma-1))
func1 = lambda M:  -M + np.sqrt((gamma+1)/(gamma-1)*(M*epsilon)**((2*(gamma-1))/(gamma+1))-2/(gamma-1))
result = root_scalar(func1,x0=3,method='newton')
Me = result.root

PePc = (1+(gamma-1)/2*Me**2)**(-gamma/(gamma-1))
print(f'Pe/Pc = {round(PePc,4)}')

CF = Gamma*np.sqrt(2*gamma/(gamma-1)*(1 - PePc**((gamma-1)/gamma))) + (PePc - Pa/Pc)*epsilon
print(f'CF = {round(CF,4)}')

At = F/(Pc*CF*144)
Dt = np.sqrt(4*At/np.pi)
print(f'At = {round(At,4)} ft^2, Dt = {round(Dt,4)} ft')

mdot = Pc*At/(cstar*etac)*144
print(f'mdot = {round(mdot,4)} lbm/s')
```
