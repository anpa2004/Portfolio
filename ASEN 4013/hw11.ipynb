```python
def calc_mixture_density(densities,weights):
    """ 
    This function calculates the density of a mixture of several different 
    components
    """
    if sum(weights)!=1:
        print('Error- Weights do not sum to 1')
    if len(densities) != len(weights):
        print('Error- Weights and chemicals do not have the same number of entries')

    sumation = 0
    for i in range(len(densities)):
        sumation += weights[i]/densities[i]
    return sumation**(-1)

# Finding mixture density
# NH4ClO4, C4H6, Zn, C
densities = [0.0705,0.14,0.258,0.00839] #lb/in^3
weights = [0.745,0.18,0.07,0.005]
rho = calc_mixture_density(densities,weights)

# Assumed Values
n = 0.2 # Burning rate exponent
Pa = 14.7 # psia
Pc0 = Pa
a = 0.00264 # in/s I think

# Given Values
hg = 1.915 # in
dg = 0.87 # dg
wg = 0.032 # in
```
