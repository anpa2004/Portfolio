```python
import numpy as np

#Givens
Pc = 225
epsilon = 62
Tc = 2800
De = 2.15
Me = 5.3
gamma = 1.3
R = 0.317
Gamma = 0.6673
Pa = 101325 #Pa

# Areas
Ae = np.pi*De**2/4
At = Ae/epsilon
print(f'Ae = {Ae} m^2, At = {At} m^2')

# mass flow
Pc_se = Pc*Pa #Pa
mdot = Pc_se*At*Gamma/(np.sqrt(1000*R*Tc))
print(f'mdot = {mdot} kg/s')

# Pressure and temp at exit
#From GASTAB
PePc = 7.80619e-4
Pe = Pc * PePc
TeTc = 0.19181
Te = Tc*TeTc
print(f'Pe = {Pe} atm, Te = {Te} K')

# Exit Velocity
Ve = np.sqrt(2*gamma/(gamma-1)*R*1000*Tc*(1-PePc**((gamma-1)/gamma)))
print(f'Ve = {Ve} m/s')

# Thrust
C = Ve + (Pe-Pa)*Ae/mdot
F = mdot*C
print(f'C = {C}, F = {F/1000} kN')




```
