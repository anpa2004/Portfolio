```python
# Problem 2 (11.20)
import numpy as np
from numpy import pi
# Givens
md = 150 #lbm/s
Pt8 = 25 #psia
Tt8 = 3600 #R
A9A8 = 1.8
g = 1.3
R = 53.34 #ftlbf/lbmR
Pt9Pt8 = 0.98
CD = 0.98
P0 = 3 #psia
gc = 32.174 #units

# Dimensions --------------------------------------------------------
M = 1
MFP = np.sqrt(g*gc/R)*M*(1+(g-1)/2*M**2)**((-g-1)/(2*g-2))
A8 = md*np.sqrt(Tt8)/(Pt8*CD*MFP)
D8 = np.sqrt(4*A8/pi)
print(f'A8 = {A8}, D8 = {D8}')

A9 = A9A8*A8
D9 = np.sqrt(4*A9/pi)
print(f'A9 = {A9}, D9 = {D9}')

# Cfg ---------------------------------------------------------------
# Finding Ideal Velocity
AAs9i = A9/(CD*A8)
print(f'(A/A*)9i = {AAs9i}')
M9i = 2.056 # FROM GASTAB @(A/A*)9i
PPt9i = 0.11853 # FROM GASTAB @ (A/A*)9i
P9i = PPt9i*Pt8
V9i = np.sqrt(R*gc*Tt8)*np.sqrt(2*g/(g-1)*(1-(P9i/Pt8)**((g-1)/g)))
print(f'From GASTAB Values, P9i = {P9i}, V9i = {V9i}')

# Finding Real Velocity
AAs9 = Pt9Pt8*A9A8/CD
print(f'(A/A*)9 = {AAs9}')
M9 = 2.0339
PPt9 = 0.12282
PtP9 = PPt9**-1
P9 = PPt9*Pt8
V9 = np.sqrt(R*gc*Tt8)*np.sqrt(2*g/(g-1)*(1-(P9/Pt8)**((g-1)/g)))
print(f'From GASTAB Values, P9 = {P9}, V9 = {V9}')

# CV
CV = V9/V9i
print(f'CV = {CV}')

# Cfg
Cfg = CD*CV*np.sqrt((1-(P9i/Pt8)**((g-1)/g))/(1-(P0/Pt8)**((g-1)/g)))*(1 + (g-1)/(2*g)*(1-P0/P9)/((PtP9)**((g-1)/g)-1))
print(f'Cfg = {Cfg}')

# Fg
Fg_actual = md*V9/gc + (P9-P0)*A9
Fg_ideal = Fg_actual / Cfg
print(f'Fg_actual = {Fg_actual}, Fg_ideal = {Fg_ideal}')
```
