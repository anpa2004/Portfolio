```python
# Givens
V0 = 0
P0 = 14.696 #psia
T0 = 518.7 #R
mdotc = 247 #lbm/s
mdotb = 1248 #lbm/s
Vec = 1190 #ft/s
Veb = 885 #ft/s
mdotf = 15750 #lbm/h
hpr = 18400 #Btu/lbm
gc = 32.174 #ftlbm/lbfs^2

# Unit Corrections
hpr = hpr*778.16 
mdotf = mdotf/3600

# a) Find thrust
Fb = mdotb*(Veb - V0)/gc
Fc = ((mdotc + mdotf)*Vec - mdotc*V0)/gc
F = Fb + Fc
print(f'Fb = {Fb}, Fc = {Fc}')
print(f'F = {F}')

# b) Find thermal efficiency
Woutc = 1/(2*gc)*((mdotc+mdotf)*Vec**2 - mdotc*V0) 
Woutb = 1/(2*gc)*((mdotb)*Veb**2 - mdotb*V0)
Wout = Woutb + Woutc
print(f'Woutb = {Woutb}, Woutc = {Woutc}')
print(f'Wout = {Wout}')
Qdot = mdotf*hpr
print(f'Qdot = {Qdot}')
etath = Wout/Qdot
print(f'eta_th = {etath}')
```
