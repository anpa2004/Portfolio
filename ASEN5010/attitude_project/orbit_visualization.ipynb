```python
import plotly.graph_objects as go
import numpy as np
from misc_functions import *
from attitude import Attitude

# Orbit Data
lmo = Attitude('lmo')
gmo = Attitude('gmo')

r_gmo = np.zeros((6501,3))
r_lmo = np.zeros((6501,3))
t_list = np.linspace(0,6500,6501)
for i,t in enumerate(t_list):
    r_gmo[i] = gmo.orbit.r_vec(t)
    r_lmo[i] = lmo.orbit.r_vec(t)

xl,yl,zl = vector_extract(r_lmo)
xg,yg,zg = vector_extract(r_gmo)

def create_mars_sphere(radius=3389.5, center=(0, 0, 0), resolution=100):
    """Generates coordinates for a Mars sphere."""
    # Create spherical coordinate grid
    u = np.linspace(0, 2 * np.pi, resolution)
    v = np.linspace(0, np.pi, resolution)
    
    # Calculate Cartesian coordinates
    x = center[0] + radius * np.outer(np.cos(u), np.sin(v))
    y = center[1] + radius * np.outer(np.sin(u), np.sin(v))
    z = center[2] + radius * np.outer(np.ones(np.size(u)), np.cos(v))
    
    return x, y, z

# Generate Mars mesh
x, y, z = create_mars_sphere()

# Create the plot
fig = go.Figure(data=[go.Mesh3d(
    x=x.flatten(), 
    y=y.flatten(), 
    z=z.flatten(), 
    color='#c1440e', # Mars red
    opacity=1.0,
    name='Mars'
)])

# Update layout for proper 3D aspect ratio
fig.update_layout(
    scene=dict(
        xaxis=dict(nticks=4, range=[-5000, 5000]),
        yaxis=dict(nticks=4, range=[-5000, 5000]),
        zaxis=dict(nticks=4, range=[-5000, 5000]),
        aspectmode='cube'
    )
)

fig.show()

# Create the 3D scatter plot
fig = go.Figure(data=[go.Scatter3d(
    x=xl, y=yl, z=zl,
    mode='lines',
    marker=dict(
        size=8,
        color=z,                # Color based on z-axis
        colorscale='Viridis',   # Choose a colorscale
        opacity=0.8
    )
), go.Mesh3d(
    x=x.flatten(), 
    y=y.flatten(), 
    z=z.flatten(), 
    color='#c1440e', # Mars red
    opacity=1.0,
    name='Mars'
)])

# Optimize layout
fig.update_layout(margin=dict(l=0, r=0, b=0, t=0))
fig.update_layout(
    scene=dict(
        xaxis=dict(nticks=4, range=[-5000, 5000]),
        yaxis=dict(nticks=4, range=[-5000, 5000]),
        zaxis=dict(nticks=4, range=[-5000, 5000]),
        aspectmode='cube'
    )
)

fig.show()
```

```python
import numpy as np
import plotly.graph_objects as go
from PIL import Image


def plot_mars_3d(
    trajectories,
    title="Mars Orbit Visualization",
    mars_radius=3390,  # km
    mars_texture_path=None,
    trajectory_labels=None,
    trajectory_colors=None,
    show_mars=True,
    show_axes=False
):
    """
    Plot Mars and 3D trajectories around it.

    Parameters
    ----------
    trajectories : array-like OR list of array-like
        Each trajectory should be shape (N,3) → [x,y,z] in km.
        Can pass a single trajectory or a list of trajectories.

    title : str
        Plot title

    mars_radius : float
        Radius of Mars (default 3390 km)

    mars_texture_path : str or None
        Path to Mars texture image (optional)

    trajectory_labels : list of str
        Labels for each trajectory

    trajectory_colors : list of str
        Plotly color strings

    show_mars : bool
        Toggle Mars sphere

    show_axes : bool
        Toggle axis visibility
    """

    # --- Ensure trajectories is a list ---
    if isinstance(trajectories, np.ndarray):
        trajectories = [trajectories]

    n_traj = len(trajectories)

    # Defaults
    if trajectory_labels is None:
        trajectory_labels = [f"Trajectory {i+1}" for i in range(n_traj)]

    if trajectory_colors is None:
        trajectory_colors = [None]*n_traj

    fig = go.Figure()

    # ============================================================
    # 🌍 Mars Sphere
    # ============================================================
    if show_mars:
        n = 100
        theta = np.linspace(0, 2*np.pi, n)
        phi = np.linspace(0, np.pi, n)
        theta, phi = np.meshgrid(theta, phi)

        x = mars_radius * np.sin(phi) * np.cos(theta)
        y = mars_radius * np.sin(phi) * np.sin(theta)
        z = mars_radius * np.cos(phi)

        if mars_texture_path is not None:
            # Load image
            img = np.asarray(Image.open(mars_texture_path))

            # Resize to match sphere
            from skimage.transform import resize
            img = resize(img, (n, n), anti_aliasing=True)

            # Build colorscale (Plotly workaround)
            surfacecolor = np.arange(n*n).reshape(n, n)

            flat_img = img.reshape(-1, 3)
            colorscale = [
                [i/(len(flat_img)-1),
                 f'rgb({int(r*255)}, {int(g*255)}, {int(b*255)})']
                for i, (r, g, b) in enumerate(flat_img)
            ]

            mars_surface = go.Surface(
                x=x, y=y, z=z,
                surfacecolor=surfacecolor,
                colorscale=colorscale,
                showscale=False,
                name="Mars",
                hoverinfo='skip'
            )

        else:
            # Simple Mars-like coloring (fast + clean)
            mars_surface = go.Surface(
                x=x, y=y, z=z,
                colorscale='Reds',
                showscale=False,
                name="Mars",
                hoverinfo='skip',
                lighting=dict(ambient=0.7, diffuse=0.8)
            )

        fig.add_trace(mars_surface)

    # ============================================================
    # 🛰️ Trajectories
    # ============================================================
    for i, traj in enumerate(trajectories):
        traj = np.asarray(traj)

        fig.add_trace(go.Scatter3d(
            x=traj[:, 0],
            y=traj[:, 1],
            z=traj[:, 2],
            mode='lines',
            name=trajectory_labels[i],
            line=dict(
                width=4,
                color=trajectory_colors[i]
            )
        ))

    # ============================================================
    # 🎨 Layout
    # ============================================================
    fig.update_layout(
        title=title,
        scene=dict(
            aspectmode='data',
            xaxis=dict(visible=show_axes),
            yaxis=dict(visible=show_axes),
            zaxis=dict(visible=show_axes),
        ),
        margin=dict(l=0, r=0, t=40, b=0)
    )

    fig.show()

plot_mars_3d([r_lmo,r_gmo])
```
