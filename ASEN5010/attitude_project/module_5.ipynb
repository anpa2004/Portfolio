### Task 8: Sun Pointing Control

```python
from attitude import Attitude
from plotting_functions import plot_2d
from misc_functions import vector_extract,create_submission_txt
from reference_frames import Reference
import numpy as np

# Setup
lmo = Attitude('lmo')
X0 = np.hstack((lmo.mrp_bn_0,lmo.omega_bn_0))
tspan = [0,500]

# Control setup
control_schedule = [{'ref':'sun','t_start':0,'t_end':tspan[-1],'K':.1,'P':1}]
lmo.control_schedule = control_schedule

# Integrating
attitude_simulated_controlled = lmo.attitude_propagation(X0,tspan,dt=1)
# Plotting
sxc,syc,szc = vector_extract(attitude_simulated_controlled['sigma'])
tc = attitude_simulated_controlled['time']
name = ['sx_uncontrolled','sy_uncontrolled','sz_uncontrolled','sx_controlled','sy_controlled','sz_controlled']
plot_2d(y=[sxc,syc,szc],x=[tc,tc,tc],line_name=name,title='Simulated Omega',xlabel='Time (mins)',ylabel='sigma')

# Checking the truth
lmo_ref = Reference('lmo')
sun = lmo_ref.sun_pointing_ref()
print(lmo.dcms.dcm_to_mrp(sun['RN']))
```
