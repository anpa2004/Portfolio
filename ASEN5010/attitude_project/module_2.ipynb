## Task 1: Orbit Simulation
#### Return the LMO state at $t=450$ seconds and GMO state at $t=1150$ seconds

```python
from orbits import Mars_Orbit
import numpy as np
import plotly.graph_objects as go
from misc_functions import vector_extract,create_submission_txt
from plotting_functions import *
lmo = Mars_Orbit('LMO')
gmo = Mars_Orbit('GMO')

# Reading state
state_450 = lmo.full_orbital_state(450,True)
print(' ')
state_1150 = gmo.full_orbital_state(1150,True)

# Creating submission text files
create_submission_txt('submission_files/m2_t1_rlmo_450.txt',state_450['r_vec'])
create_submission_txt('submission_files/m2_t1_vlmo_450.txt',state_450['v_vec'])
create_submission_txt('submission_files/m2_t1_rgmo_1150.txt',state_1150['r_vec'])
create_submission_txt('submission_files/m2_t1_vgmo_1150.txt',state_1150['v_vec'])

# Creating propagated graph for submission
t0 = 0
tf = 4*7200 # Seconds (8hrs)
N = int(tf/10) # num points

t_list = np.linspace(t0,tf,N+1)
vecs_lmo = np.zeros((len(t_list),3))
vecs_gmo = np.zeros((len(t_list),3))

for t in t_list:
    state_l = lmo.full_orbital_state(t,False)
    state_g = gmo.full_orbital_state(t,False)

    vecs_lmo[int(t/10)] = state_l['r_vec']
    vecs_gmo[int(t/10)] = state_g['r_vec']

x_l,y_l,z_l = vector_extract(vecs_lmo)
x_g,y_g,z_g = vector_extract(vecs_gmo)
r_l = [np.linalg.norm(vec) for vec in vecs_lmo]
r_g = [np.linalg.norm(vec) for vec in vecs_gmo]
t_list = [t/3600 for t in t_list]
x = [t_list,t_list,t_list,t_list]
y_lmo = [x_l,y_l,z_l,r_l]
y_gmo = [x_g,y_g,z_g,r_g]

lines = ['x','y','z','Distance']

plot_2d(y_lmo,x=x,xlabel='Time (hr)',ylabel='Position (km)',title = 'LMO Satellite Position Propagation',line_name=lines)
plot_2d(y_gmo,x=x,xlabel='Time (hr)',ylabel='Position (km)',title = 'GMO Satellite Position Propagation',line_name=lines)
```
