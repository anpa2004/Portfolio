```python
from dcms import Attitude
from numerical import Integrator
from dcms import Linear
import numpy as np
import matplotlib.pyplot as plt
at = Attitude()
ln = Linear()
ig = Integrator()

def f(t,q):
    """   
    Function to integrate
    """
    omega = 3*np.array([np.sin(0.1*t),0.01,np.cos(0.1*t)])
    omega = [o*np.pi/180 for o in omega]
    qdot = 0.5*(np.eye(3)+ln.tilde(q) + np.outer(q,q))@ omega
    return qdot

q0 = np.array([0.4,0.2,-0.1])
tspan = [0,42]
t,q = ig.ode45(f,q0,tspan)
final = q[:,-1]
plt.plot(t,np.transpose(q))
plt.scatter(42,final[0])
plt.scatter(42,final[1])
plt.scatter(42,final[2])
plt.show()
print(np.linalg.norm(final))

```

```python
from dynamics import Dynamics
import numpy as np
dyn = Dynamics()

R1 = np.array([1,-1,2])
R2 = np.array([-1,-3,2])
R3 = np.array([2,-1,-1])
R4 = np.array([3,-1,-2])
R = [R1,R2,R3,R4]
Rd1 = np.array([2,1,1])
Rd2 = np.array([0,-1,1])
Rd3 = np.array([3,2,-1])
Rd4 = np.array([0,0,1])
Rd = [Rd1,Rd2,Rd3,Rd4]

m = [1,1,2,2]

# Finding about origin
Ho = dyn.angular_momentum_discrete(R,Rd,m)
print(f'Ho = {Ho.tolist()}')

# Finding about CG
Rc = dyn.center_of_gravity_pos_discrete(R,m)
Rdc = dyn.center_of_gravity_vel_discrete(Rd,m)
Hc = dyn.angular_momentum_discrete(R,Rd,m,rp=Rc,rpd=Rdc)
print(f'Hc = {Hc.tolist()}')

# Testing if they should be the same
test_val = np.cross(Rc,sum(m)*Rdc)
print(f'should be zero = {test_val}')
```
