##### Create a graph of eccentricity vs SMA following a close flyby of earth

```python
import numpy as np
import matplotlib.pyplot as plt
a0 = 1.5 #AU
e0 = 0.5
c0 = 1/(2*a0) + np.sqrt(a0*(1-e0**2))  # Unit abomination


def post_flyby_ecc(c0,a1):
    """
    This function determines the eccentricity after a flyby of a planet
    given an initial Tisserand constant c0 and a known semimajor axis 
    after the flyby a1

    Args:
        c0: Initial Tisserand Constant
        a1: Final semimajor axis after the flyby

    Returns:
        e1: Eccentricity after the flyby
    """
    e1 = np.sqrt(1-1/a1*(c0 - 1/(2*a1))**2)
    return e1

a1 = np.linspace(0.3,10,400) # AU
e1_list = post_flyby_ecc(c0,a1)
plt.plot(a1,e1_list,label = 'Eccentricity')
plt.axvline(a0,label='a0 = 1.5 AU',color='r')
plt.xlabel('Semimajor Axis After Flyby (AU)')
plt.ylabel('Eccentricity after Flyby (Unitless)')
plt.grid()
plt.legend()
plt.title('Eccentricity After Flyby versus Semimajor Axis After Flyby')
plt.show()
print(max(e1_list))
e = post_flyby_ecc(c0,0.8905)
print(e)
e = post_flyby_ecc(c0,4.1558)
print(e)
print(min(e1_list))
```

```python
import numpy as np
mus = 1.33e11 # km^3/s^2
mup = 4e5 # km^3/s^2
R = 1.5e8 # km
a0 = 1.5*R #AU
e0 = 0.5
c0 = 1/(2*a0) + np.sqrt(a0*(1-e0**2))

# Planet Orbital Velocity
Vp = np.sqrt(mus/R)
print(f'Vp = {round(Vp,4)} km/s')

# Angular Momentum
h = np.sqrt(mus*a0*(1-e0**2))
print(f'h = {round(h,4)} km^2/s')

# Tangential and radial speed of spacecraft before planet interaction
Vt = h/R
print(f'Vt = {round(Vt,4)} km/s')
Vsc = np.sqrt(mus*(2/R - 1/a0)) # Velocity at r=R before flyby
print(f'Vsc = {round(Vsc,4)} km/s')
Vr = np.sqrt(Vsc**2 - Vt**2)
print(f'Vr = {round(Vr,4)} km/s')

# Vinfinty incoming
rhat = np.array([1,0])
thetahat = np.array([0,1])
Vinf_inp = Vr*rhat + (Vt-Vp)*thetahat
Vinf_inm = - Vr*rhat + (Vt-Vp)*thetahat
print(f'V_inf,in,p = {Vinf_inp} km/s')
Vinf_in = [Vinf_inp, Vinf_inm]

# Finding Hyperbolic Orbit Properties
print(' ')
RE = 6400 # km
Vinfty = np.linalg.norm(Vinf_inp)
eh = 1 + (RE * Vinfty**2/mup)
delta = 2*np.arcsin(1/eh)
print(f'Vinf = {round(Vinfty,4)} km/s')
print(f'eh = {round(eh,4)}')
print(f'delta = {round(delta,4)} rad = {round(delta*180/np.pi,4)} deg')

# Rotation and out velocity
print('\nPositive Delta')
Rot = np.array([[np.cos(delta),-np.sin(delta)],[np.sin(delta),np.cos(delta)]])
print(f'R = {Rot}')
Vinfout = Rot@Vinf_inp
print(f'Vinf,out = {Vinfout} km/s')
Vscout = Vp*thetahat + Vinfout
print(f'Vsc,out = {Vscout} km/s')
vscout_norm = np.linalg.norm(Vscout)
print(f'Vscout_norm = {vscout_norm} km/s')
a1max = (2/R - vscout_norm**2/mus)**-1
print(f'a1 = {round(a1max/R,4)} AU')
e1max = post_flyby_ecc(c0,a1max/R)
print(f'e1 = {round(e1max,4)}')

print('\nNegative Delta')
Rot = np.array([[np.cos(-delta),-np.sin(-delta)],[np.sin(-delta),np.cos(-delta)]])
print(f'R = {Rot}')
Vinfout = Rot@Vinf_inp
print(f'Vinf,out = {Vinfout} km/s')
Vscout = Vp*thetahat + Vinfout
print(f'Vsc,out = {Vscout} km/s')
vscout_norm = np.linalg.norm(Vscout)
print(f'Vscout_norm = {vscout_norm} km/s')
a1min = (2/R - vscout_norm**2/mus)**-1
print(f'a1 = {round(a1min/R,4)} AU')
e1min = post_flyby_ecc(c0,a1min/R)
print(f'e1 = {round(e1min,4)}')

# Plotting
a1 = np.linspace(0.3,10,400) # AU
e1 = post_flyby_ecc(c0,a1)
plt.plot(a1,e1_list,label = 'Eccentricity')
plt.scatter(a0/R,e0,label='(a0,e0)')
plt.axvline(a1max/R,label='Max a1',color='g')
plt.axvline(a1min/R,label='Min a1',color='r')
plt.xlabel('Semimajor Axis After Flyby (AU)')
plt.ylabel('Eccentricity after Flyby (Unitless)')
plt.grid()
plt.legend()
plt.title('Eccentricity After Flyby versus Semimajor Axis After Flyby')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
a1max = 4.1558 # AU
a1min = 0.8905 # AU
mus = 1.33e11 # km^3/s^2
mup = 4e5 # km^3/s^2
R = 1.5e8 # km
a0 = 1.5*R #km
e0 = 0.5

n = np.sqrt(mus/a0**3)

def post_flyby_ecc(c0,a1):
    """
    This function determines the eccentricity after a flyby of a planet
    given an initial Tisserand constant c0 and a known semimajor axis 
    after the flyby a1

    Args:
        c0: Initial Tisserand Constant
        a1: Final semimajor axis after the flyby

    Returns:
        e1: Eccentricity after the flyby
    """
    e1 = np.sqrt(1-1/a1*(c0 - 1/(2*a1))**2)
    return e1

a1 = np.linspace(a1min*R-0.5*R,a1max*R+0.5*R,100)
a1_au = [a/R for a in a1]
def post_flyby_inc(a1,a0,n,c0,mu,R):
    """   
    This function finds the inclination after a flyby based on the assumption that the 
    pojectile is initially in a polar orbit with respect to the earth's orbit. 

    Args:
        a1: Ending sma after flyby
        a0: SMA before flyby
        n: mean motion
        c0: Tisserand's constant before flyby
        mu: gravitational parameter of sun
        R: distance from earth to sun (1AU)

    Returns:
        i: Inclination after flyby
    """
    e1 = post_flyby_ecc(c0,a1/R)
    i = np.arccos(mu/2*(1/a1 - 1/a0)/(n*np.sqrt(mu*a1*(1-e1**2))))
    i = i*180/np.pi
    return i

i1 = [post_flyby_inc(a,a0,n,c0,mus,R) for a in a1]
plt.plot(a1_au,i1,label='Inclination')
plt.axvline(a1max,label='a1max',color = 'g')
plt.axvline(a1min,label='a1min',color = 'r')
plt.grid()
plt.xlabel('Semimajor Axis After Flyby (AU)')
plt.ylabel('Inclination (deg)')
plt.title('Inclination After Flyby With Earth')
plt.legend()
plt.show()
print(np.nanmax(i1))
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import fsolve 

a1max = 4.1558 # AU
a1min = 0.8905 # AU
mus = 1.33e11 # km^3/s^2
mup = 4e5 # km^3/s^2
R = 1.5e8 # km
a0 = 1.5*R #km
e0 = 0.5

n = np.sqrt(mus/a0**3)

def post_flyby_ecc_inc(a1,i1,n,mus,a0):
    """
    This function determines the eccentricity after a flyby of a planet
    given an initial Tisserand constant c0 and a known semimajor axis 
    after the flyby a1

    Args:
        c0: Initial Tisserand Constant
        a1: Final semimajor axis after the flyby
        i1: Inclination after flyby

    Returns:
        e1: Eccentricity after the flyby
    """
    # e1 = np.sqrt(1-1/a1*(c0 - 1/(2*a1))**2)
    e1 = np.sqrt(1-(mus/2)*(-1/a0 + 1/a1)**2/(n**2*np.cos(i1)**2*mus*a1))
    return e1

a1 = np.linspace(a1min*R-0.5*R,a1max*R+0.5*R,100)
a1_au = [a/R for a in a1]
def post_flyby_inc(a1,a0,n,mu,R):
    """   
    This function finds the inclination after a flyby based on the assumption that the 
    pojectile is initially in a polar orbit with respect to the earth's orbit. 

    Args:
        a1: Ending sma after flyby
        a0: SMA before flyby
        n: mean motion
        c0: Tisserand's constant before flyby
        mu: gravitational parameter of sun
        R: distance from earth to sun (1AU)

    Returns:
        i: Inclination after flyby
    """
    #e1 = post_flyby_ecc(c0,a1/R)

    ifunc = lambda i: -i + np.arccos(mu/2*(1/a1 - 1/a0)/(n*np.sqrt(mu*a1*(1-post_flyby_ecc_inc(a1,i,n,mu,a0)**2))))
    i_guess = np.pi/3
    root_result = fsolve(ifunc,i_guess)
    i = root_result[0]
    i = i*180/np.pi
    return i

i1 = [post_flyby_inc(a,a0,n,mus,R) for a in a1]
plt.plot(a1_au,i1,label='Inclination')
plt.axvline(a1max,label='a1max',color = 'g')
plt.axvline(a1min,label='a1min',color = 'r')
plt.grid()
plt.xlabel('Semimajor Axis After Flyby (AU)')
plt.ylabel('Inclination (deg)')
plt.title('Inclination After Flyby With Earth')
plt.legend()
plt.show()
```
